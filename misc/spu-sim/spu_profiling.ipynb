# 探索SPU

```python
def fun(a, b, c):
    return a * b + c

result = fun(3, 2, 1)

print(f"""
-----------------------------------------------
result = {result}
-----------------------------------------------
""")
```

## 初始化SPU集群

```python
import sml.utils.emulation as emulation

mode = emulation.Mode.MULTIPROCESS

emulator = emulation.Emulator(
    "utils/3pc.json",
    mode,
    bandwidth=100,
    latency=10
)

emulator.up()
```

## 在SPU下运行

```python
a, b, c = 3, 2, 1
a, b, c = emulator.seal(a, b, c)
result = emulator.run(fun)(a, b, c)
print(f"""
-----------------------------------------------
result = {result}
-----------------------------------------------
""")
```

## 包含子函数

```python
def fun1(x):
    return x+1

def fun2(x):
    return x*2

def fun3(x):
    x = fun1(x)
    x = fun2(x)
    return x

x = 2
x = emulator.seal(x)
```

```python
result = emulator.run(fun3)(x)
```

```python
print(result)
```

## 关闭模拟

```python
emulator.down()
```

# 探索SPU

```python
def fun(a, b, c):
    return a * b + c


result = fun(3, 2, 1)

print(f"""
-----------------------------------------------
result = {result}
-----------------------------------------------
""")
```

## 初始化SPU集群

```python
import sml.utils.emulation as emulation

mode = emulation.Mode.MULTIPROCESS

emulator = emulation.Emulator(
    "3pc.json",
    mode,
    bandwidth=100,
    latency=10
)

emulator.up()
```

## 在SPU下运行

```python
a, b, c = 3, 2, 1
a, b, c = emulator.seal(a, b, c)
result = emulator.run(fun)(a, b, c)
print(f"""
-----------------------------------------------
result = {result}
-----------------------------------------------
""")
```

## 关闭模拟

```python
emulator.down()
```
